```python
from ultralytics import YOLO

# 1. Khởi tạo model với cấu trúc P2
model = YOLO('yolo12n.pt')

# 2. Huấn luyện với các tính năng nâng cao
results =model.train(
    data="data.yaml",
    imgsz=832,
    batch=4,          
    epochs=50,       
    optimizer="AdamW",
    lr0=0.001,
    lrf=0.01,
    cos_lr=True,
    warmup_epochs=3.0, 
    mosaic=1.0,        
    mixup=0.1,        
    fliplr=0.5,
    scale=0.5,
    patience=20,
    close_mosaic=10, 
)
```
