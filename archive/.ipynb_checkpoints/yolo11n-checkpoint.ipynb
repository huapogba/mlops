```python
import cv2

img = cv2.imread("train/images/001.jpg")

h, w, _ = img.shape()

with open("train/labels/001.txt") as f:
    lines = f.readlines()

for line in lines:
    cls, x, y, bw, bh = map(float, line.split())

    x1 = int((x - bw/2) * w)
    y1 = int((y - bh/2) * h)
    x2 = int((x + bw/2) * w)
    y2 = int((y + bh/2) * h)

    color = (0,255,0) if int(cls)==1 else (255,0,0)

    cv2.rectangle(img,(x1,y1),(x2,y2),color,2)

cv2.imshow("label check", img)
cv2.waitKey(0)
```
