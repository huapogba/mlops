```python
import torch
torch.cuda.empty_cache()
```

```python
from ultralytics import YOLO

# 1. Khởi tạo model với cấu trúc P2
model = YOLO('yolov8_p2.yaml').load('yolov8s.pt') 

# 2. Huấn luyện với các tính năng nâng cao
results = model.train(
    # --- Cấu hình cơ bản ---
    data='data.yaml', 
    epochs=60, 
    imgsz=832,          
    batch=8,            
    device=0,           
    project='crowd_detect', 
    name='yolov8_p2_advanced',

    # --- 1. Augmentation (Biến đổi ảnh & Scale) ---
    scale=0.5,          # Scale ảnh (+/- 50%) giúp model học vật thể ở xa/gần
    mosaic=1.0,         # Gộp 4 ảnh thành 1 (cực kỳ tốt cho việc học vật thể nhỏ)
    mixup=0.1,          # Trộn 2 ảnh (giúp model bớt overfit)
    flipud=0.0,         # Lật ảnh theo chiều dọc (đối với người thì thường để 0.0)
    fliplr=0.5,         # Lật ảnh trái phải (0.5 = 50% xác suất)
    # --- 2. Learning Rate & Scheduler (Biến đổi theo epoch) ---
    lr0=0.01,           # Learning rate khởi đầu
    lrf=0.01,           # Learning rate kết thúc (lr0 * lrf)
    cos_lr=True,        # Sử dụng Cosine Annealing scheduler (giảm LR theo hình sin)
    warmup_epochs=3.0,  # 3 epoch đầu sẽ chạy chậm để ổn định weight
    # --- 3. Early Stopping ---
    patience=10,        # Nếu sau 10 epoch mà mAP không tăng, model sẽ tự dừng
    # --- Tối ưu thêm cho P2 ---
    overlap_mask=True,  # Hỗ trợ xử lý các bounding box đè lên nhau trong đám đông
)
```
